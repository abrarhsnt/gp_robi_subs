```python
import re, unicodedata
import pandas as pd


# 1. PASTE the BTRC table here (copied from the browser)

RAW_TEXT = """
মাস	গ্রামীণফোন (মিলিয়ন)	রবি আজিয়াটা (মিলিয়ন)	বাংলালিংক (মিলিয়ন)	টেলিটক (মিলিয়ন)	মোট গ্রাহক (মিলিয়ন)
আগস্ট, ২০২৬

৮৭.২০

৫৯.০৪

৩৭.৯৫

৬.৯০

১৯১.০৯

জুলাই, ২০২৬

৮৭.০১

৫৮.৭৭

৩৭.৮৪

৬.৮২

১৯০.৪৪

জুন, ২০২৬

৮৬.৫৯

৫৮.৫৭

৩৭.৮৬

৬.৮০

১৮৯.৮৩

মে, ২০২৬

৮৫.৯০

৫৮.১৮

৩৭.৭১

৬.৮১

১৮৮.৬০

এপ্রিল, ২০২৬

৮৪.৯৬

৫৭.৮০

৩৭.৪৯

৬.৮২

১৮৭.০৭

মার্চ, ২০২৬

৮৪.৮৪

৫৭.৩৯

৩৭.৩৭

৬.৮২

১৮৬.০৬

ফেব্রুয়ারী, ২০২৬

৮৪.৩৬

৫৭.৩০

৩৭.৩৫

৬.৮২

১৮৫.৮৪

জানুয়ারী, ২০২৬

৮৪.৩৩

৫৭.২৪

৩৭.৪০

৬.৮৩

১৮৫.৮০

ডিসেম্বর, ২০২৫

৮৪.১৬

৫৭.৪০

৩৭.৫২

৬.৮০

১৮৫.৮৯

নভেম্বর, ২০২৫	৮৫.০১	৫৭.৫৭	৩৭.৭০	৬.৭৮	১৮৭.০৬
অক্টোবর, ২০২৫	৮৫.৩৯	৫৭.৫৪	৩৭.৮১	৬.৭২	১৮৭.৪৬
সেপ্টেম্বর, ২০২৫	৮৫.৮৫	৫৭.৫২	৩৭.৯৩	৬.৬৭	১৮৭.৯৭
অগাস্ট, ২০২৫	৮৬.৪৭	৫৭.৪১	৩৮.০৫	৬.৬৪	১৮৮.৫৭
জুলাই, ২০২৫	৮৬.৬৪	৫৭.৫৪	৩৮.০৮	৬.৬১	১৮৮.৮৭
জুন, ২০২৫	৮৬.৫১	৫৭.৪০	৩৭.৯৫	৬.৫৯	১৮৮.৪৫
মে, ২০২৫	৮৬.১৭	৫৬.৯৮	৩৭.৮৮	৬.৫৮	১৮৭.৬১
এপ্রিল, ২০২৫	৮৫.৪৮	৫৬.৬০	৩৮.০০	৬.৫৬	১৮৬.৬৪
মার্চ, ২০২৫	৮৫.০৯	৫৬.৩৬	৩৮.২৩	৬.৫৪	১৮৬.২২
ফেব্রুয়ারী, ২০২৫	৮৪.৯০	৫৬.৩৯	৩৮.৬৪	৬.৫৬	১৮৬.৫০
জানুয়ারি, ২০২৫	৮৪.৫৮	৫৬.৪০	৩৯.০৫	৬.৫৭	১৮৬.৫৯
ডিসেম্বর, ২০২৪	৮৪.৫২	৫৬.৭৩	৩৯.৭৭	৬.৫৬	১৮৭.৫৮
নভেম্বর, ২০২৪	৮৪.৫৫	৫৭.১৪	৪০.৫৩	৬.৫৬	১৮৮.৭৮
অক্টোবর ২০২৪	৮৪.৭৪	৫৭.৬০	৪১.০৭	৬.৫৫	১৮৯.৯৬
সেপ্টেম্বর ২০২৪	৮৪.৮৩	৫৭.৮৮	৪১.৬৩	৬.৫২	১৯০.৮৬
আগষ্ট ২০২৪	৮৫.০৮	৫৮.৩৬	৪২.৪৭	৬.৫৩	১৯২.৪৩
জুলাই ২০২৪	৮৫.২৪	৫৮.৯২	৪৩.৫৬	৬.৫৫	১৯৪.২৬
জুন ২০২৪	৮৫.৫৩	৫৯.৫১	৪৪.৪৮	৬.৫৬	১৯৬.০৮
মে ২০২৪	৮৪.৯১	৫৮.৮৮	৪৪.৭৩	৬.৫৭	১৯৫.০৯
এপ্রিল ২০২৪	৮৩.৯৫	৫৮.৫১	৪৪.৭২	৬.৫৫	১৯৩.৭৩
মার্চ ২০২৪	৮৩.২২	৫৮.০৭	৪৪.৪৩	৬.৫৩	১৯২.২৬
ফেব্রুয়ারি ২০২৪	৮২.৫৮	৫৮.৩৪	৪৩.৯৬	৬.৫০	১৯১.৩৮
জানুয়ারি ২০২৪	৮২.২৭	৫৮.২৭	৪৩.৪৪	৬.৪৮	১৯০.৪৬
ডিসেম্বর ২০২৩	৮২.২০	৫৮.৬৭	৪৩.৪৮	৬.৪৬	১৯০.৮১
নভেম্বর ২০২৩	৮২.১৪	৫৮.৩৮	৪৩.৩৮	৬.৪৬	১৯০.৩৬
অক্টোবর ২০২৩	৮২.১৯	৫৭.৯৬	৪৩.০৯	৬.৪৩	১৮৯.৬৭
সেপ্টেম্বর ২০২৩	৮২.২২	৫৭.৬২	৪২.৮৫	৬.৪৪	১৮৯.১৩
আগস্ট ২০২৩	৮২.১৪	৫৭.৩৩	৪২.৭২	৬.৪৫	১৮৮.৬৪
জুলাই ২০২৩	৮১.৮৯	৫৬.৮২	৪২.৩০	৬.৪৭	১৮৭.৪৮
জুন ২০২৩	৮১.৪৫	৫৬.৩৫	৪১.৮১	৬.৪৯	১৮৬.১০
মে ২০২৩	৮১.২৪	৫৫.৯২	৪১.৪২	৬.৫৪	১৮৫.১৩
এপ্রিল ২০২৩	৮০.৫৯	৫৫.৭২	৪১.৩২	৬.৫৭	১৮৪.২০
মার্চ ২০২৩	৮০.৩০	৫৫.৫৭	৪১.৩৯	৬.৬৩	১৮৩.৮৯
ফেব্রুয়ারি ২০২৩	৭৯.৯৫	৫৫.১৪	৪০.৮৫	৬.৬৭	১৮২.৬১
জানুয়ারি ২০২৩	৭৯.৩৭	৫৪.৬১	৪০.১৪	৬.৬৮	১৮০.৮০
ডিসেম্বর ২০২২	৭৯.২৯	৫৪.৪০	৩৯.৮১	৬.৬৯	১৮০.২০
নভেম্বর ২০২২	৮০.৩৪	৫৪.৩৪	৩৯.৪৬	৬.৭৩	১৮০.৮৭
অক্টোবর ২০২২	৮১.৪১	৫৪.২৮	৩৯.২৩	৬.৭৫	১৮১.৬৭
সেপ্টেম্বর ২০২২	৮১.৯৫	৫৪.৩৬	৩৮.৩৪	৬.৭৮	১৮১.৪৩
আগস্ট ২০২২	৮৩.১৫	৫৪.৯৫	৩৮.৭৬	৬.৭২	১৮৩.৫৮
জুলাই ২০২২	৮৪.০৮	৫৪.৭৭	৩৮.৪৮	৬.৭১	১৮৪.০৫
জুন ২০২২	৮৪.৮০	৫৪.৫৩	৩৮.৩৭	৬.৭৫	১৮৪.৪৫
মে ২০২২	৮৪.৯৫	৫৪.২৫	৩৮.২৫	৬.৭৮	১৮৪.২৩
এপ্রিল ২০২২	৮৪.২৭	৫৪.০০	৩৮.২৫	৬.৮৬	১৮৩.৩৮
মার্চ ২০২২	৮৩.৮৭	৫৪.০৭	৩৮.০৯	৬.৮৯	১৮২.৯২
ফেব্রুয়ারি ২০২২	৮৩.১৮	৫৩.৭৮	৩৭.৭০	৬.৮৮	১৮১.৫৪
জানুয়ারি ২০২২	৮৩.০২	৫৩.৫৭	৩৭.৪১	৬.৭৮	১৮০.৭৮
ডিসেম্বর ২০২১	৮৩.৪৬	৫৩.৬৭	৩৭.২২	৬.৬৭	১৮১.০২
নভেম্বর ২০২১	৮৪.০৪	৫৩.৭২	৩৭.২১	৬.৫৬	১৮১.৫৩
অক্টোবর ২০২১	৮৪.১২	৫৩.৫৪	৩৭.১৭	৬.৪৯	১৮১.৩২
সেপ্টেম্বর ২০২১	৮৩.৭৮	৫৩.০৪	৩৭.০২	৬.৪১	১৮০.২৫
আগস্ট ২০২১	৮৩.১৮	৫২.২৬	৩৬.৯০	৬.২৭	১৭৮.৬১
"""


# 2. Parse the pasted table

def clean(s):
    s = unicodedata.normalize("NFC", s)
    return re.sub(r"[\u200b\u200c\u200d\ufeff]", "", s)

BN_DIGITS = str.maketrans("০১২৩৪৫৬৭৮৯", "0123456789")
_M = {
    "জানুয়ারি": 1, "জানুয়ারী": 1, "ফেব্রুয়ারি": 2, "ফেব্রুয়ারী": 2,
    "মার্চ": 3, "এপ্রিল": 4, "মে": 5, "জুন": 6, "জুলাই": 7,
    "আগস্ট": 8, "অগাস্ট": 8, "আগষ্ট": 8, "অগাষ্ট": 8,
    "সেপ্টেম্বর": 9, "সেপ্টেম্বার": 9, "অক্টোবর": 10, "অক্টোবার": 10,
    "নভেম্বর": 11, "নভেম্বার": 11, "ডিসেম্বর": 12, "ডিসেম্বার": 12,
    "january": 1, "february": 2, "march": 3, "april": 4, "may": 5, "june": 6, "july": 7,
    "august": 8, "september": 9, "october": 10, "november": 11, "december": 12,
}
MONTHS = {clean(k).lower(): v for k, v in _M.items()}
OPS = ["grameenphone", "robi", "banglalink", "teletalk"]
COLUMNS = ["month"] + OPS + ["total", "source"]
MONTH_RE = re.compile(r"^\s*([^\d,]+?)\s*,?\s*(\d{4})\s*$")
NUM_RE = re.compile(r"^-?\d[\d,]*\.?\d*$")

def parse_btrc_text(text):
    text = clean(text).translate(BN_DIGITS)
    tokens = [t.strip() for t in re.split(r"[\t\r\n]+|\s{2,}", text)]
    tokens = [t for t in tokens if t]
    records, current = [], None
    for tok in tokens:
        m = MONTH_RE.match(tok)
        if m:
            current = {"name": m.group(1).strip(), "year": m.group(2), "nums": []}
            records.append(current)
        elif NUM_RE.match(tok) and current is not None:
            current["nums"].append(float(tok.replace(",", "")))
    out = []
    for r in records:
        month_no = MONTHS.get(r["name"].lower())
        if not month_no:
            print("Skipped (month name not recognised):", r["name"], r["year"])
        elif len(r["nums"]) != 5:
            print(f"Skipped ({len(r['nums'])} numbers instead of 5):", r["name"], r["year"], r["nums"])
        else:
            out.append((f"{r['year']}-{month_no:02d}", *r["nums"], "BTRC mobile subscribers page"))
    return pd.DataFrame(out, columns=COLUMNS)

btrc = parse_btrc_text(RAW_TEXT)
btrc = btrc.drop_duplicates("month", keep="first")
print(len(btrc), "months parsed")


# 3. Validate

df = btrc.copy()
df["month"] = pd.to_datetime(df["month"], format="%Y-%m")
df = df.sort_values("month").reset_index(drop=True)

df["diff"] = (df[OPS].sum(axis=1) - df["total"]).abs()
bad = df[df["diff"] > 0.15]
print("Rows where operators don't add up to the total (>0.15m):",
      "none" if bad.empty else "")
if not bad.empty:
    print(bad[["month", "total", "diff"]].to_string(index=False))

full = pd.date_range(df["month"].min(), df["month"].max(), freq="MS")
missing = full.difference(df["month"])
print(f"Months covered: {len(df)} of {len(full)} "
      f"({df['month'].min():%Y-%m} to {df['month'].max():%Y-%m})")
if len(missing):
    print("Missing months:", [d.strftime("%Y-%m") for d in missing])


# 4. Events (add only real, sourced ones)

events = pd.DataFrame([("2022-06", "BTRC bans SIM sales by Grameenphone")],
                      columns=["month", "event_label"])
events["month"] = pd.to_datetime(events["month"], format="%Y-%m")


# 5. Long format + metrics

NAMES = {"grameenphone": "Grameenphone", "robi": "Robi Axiata",
         "banglalink": "Banglalink", "teletalk": "Teletalk"}

long = df.melt(id_vars=["month", "total", "source"], value_vars=OPS,
               var_name="operator_key", value_name="subscribers_m")
long["operator"] = long["operator_key"].map(NAMES)
long = long.dropna(subset=["subscribers_m"]).rename(columns={"total": "industry_total_m"})
long = long.sort_values(["operator", "month"]).reset_index(drop=True)

lookup = long.set_index(["operator", "month"])["subscribers_m"]
def shifted(months):
    keys = pd.MultiIndex.from_arrays([long["operator"], long["month"] - pd.DateOffset(months=months)])
    return lookup.reindex(keys).to_numpy()

long["mom_change_m"] = (long["subscribers_m"] - shifted(1)).round(3)
long["mom_pct"] = ((long["subscribers_m"] / shifted(1) - 1) * 100).round(2)
long["yoy_change_m"] = (long["subscribers_m"] - shifted(12)).round(3)
long["yoy_pct"] = ((long["subscribers_m"] / shifted(12) - 1) * 100).round(2)
long["market_share_pct"] = (long["subscribers_m"] / long["industry_total_m"] * 100).round(2)
long["subscribers"] = (long["subscribers_m"] * 1_000_000).round().astype("int64")
long["is_focus_rival"] = long["operator"].isin(["Grameenphone", "Robi Axiata"])
long["year"] = long["month"].dt.year
long = long.merge(events, on="month", how="left")

out_cols = ["month", "year", "operator", "is_focus_rival", "subscribers", "subscribers_m",
            "industry_total_m", "market_share_pct", "mom_change_m", "mom_pct",
            "yoy_change_m", "yoy_pct", "event_label", "source"]
final = long[out_cols].sort_values(["month", "operator"]).reset_index(drop=True)
final["month"] = final["month"].dt.strftime("%Y-%m-%d")

#
# 6. Export
#
OUT = "gp_vs_robi_telecom.csv"
final.to_csv(OUT, index=False, encoding="utf-8")
print(f"Saved {OUT}: {len(final)} rows, {final['operator'].nunique()} operators, "
      f"{final['month'].nunique()} months")
try:
    from google.colab import files
    files.download(OUT)
except ImportError:
    pass
final.tail(8)
```
